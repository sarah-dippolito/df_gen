```python
# use case 1 (college)

from dataframe_generator import df_gen
import numpy as np
import pandas as pd 

non_us_countries = [
    "Canada", "Mexico", "India", "China", "United Kingdom",
    "Germany", "France", "Brazil", "Nigeria", "South Korea"
]

gen = df_gen(10, random_state=10)
gen.add_column("id", ("prefix_id", "STU", 5, True))
gen.add_column("names", ("names",))
gen.add_column("student_email", ("email", ["school.edu"]))
gen.add_column("phone", ("phone",[813,833,767]))
gen.add_column("state", ("state", {"missing": 0.25}))
gen.add_column(
    "country",
    (
        "computed",
        lambda df: pd.Series(
            np.where(
                df["state"].notna(),
                "United States",
                np.random.choice(non_us_countries, size=len(df))
            ),
            dtype="object"
        ),
        ["state"]
    )
)
gen.add_column("age", ("int", 18, 22))
gen.add_column("dob", ("dob_from_age", "age"))
gen.add_column("major", ("category", ["art", "math", "music", "biology", "chemistry", "english", "political science", "psychology"]))
gen.add_column("GPA", ("normal_clip", 3,1.5, 0,4, {"round": 2}))
gen.add_column("major_GPA", ("decimals", 2, 4))
gen.add_column(
    "year",
    (
        "computed",
        lambda df: pd.Series(
            np.select(
                [
                    df["age"].between(18, 19),
                    df["age"].between(19, 20),
                    df["age"].between(20, 21),
                ],
                [
                    "Freshman",
                    "Sophomore",
                    "Junior",
                ],
                default="Senior"
            ),
            dtype="object"
        ),
        ["age"]
    )
)
df = gen.build()

df.head(10)
```

```python
# use case 2 (orders)

from dataframe_generator import df_gen
import numpy as np
import pandas as pd 

gen_df = df_gen(rows=10, random_state=2)


gen_df.add_column("order_id", ("id", 0))
gen_df.add_column("customer_code", ("prefix_id", "CUST", 4, True))
gen_df.add_column("customer_id", ("alphanumeric_id", 8, False))
gen_df.add_column("customer_initials", ("string", 3, "upper"))
gen_df.add_column("num_products", ("int", 1, 20))
gen_df.add_column("price", ("correlated", "num_products", 3, 0.5))
gen_df.add_column("shipping", ("int_list", [1,1.5,2], [0.25,0.5,0.25]))
gen_df.add_column("delivered", ("bernoulli", 0.65))
gen_df.add_column("pct_paid", ("percent", 10,2, True))
gen_df.add_column("num_payment_methods", ("int_list", [1,2,3], [0.9, 0.06, 0.04]))
gen_df.add_column("order_date", ("datetime", '2025-01-01', '2025-12-31'))
gen_df.add_column("delivered_from", ("category_prob", ['Amazon', 'Website', 'Store'], [0.4, 0.5, 0.1]))
gen_df.add_column("survey_completion", ("beta", 3, 5, 100, {'missing':0.7}))  #missing used here
df = gen_df.build()

df.head(10)
```

```python
# use case 1 (all column types)
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", None)


from dataframe_generator import df_gen

gen = (
    df_gen(rows=50, random_state=14)

    
    .add_column("row_id", ("id",))
    .add_column("user_code", ("prefix_id", "USR", 4, True))
    .add_column("session_id", ("alphanumeric_id", 8))
    .add_column("group", ("manual", ["A", "B", "C"], True))
    .add_column("age", ("int", 18, 65))
    .add_column("children", ("int_list", [0, 1, 2, 3], [0.4, 0.3, 0.2, 0.1]))
    .add_column("score_uniform", ("uniform", 0, 100, {"round": 1}))
    .add_column("completion_rate", ("percent", 2, 5, True))
    .add_column("is_active", ("bool",))
    .add_column("converted", ("bernoulli", 0.3))
    .add_column("plan", ("category", ["Free", "Standard", "Premium"]))
    .add_column("region", ("category_prob",
        ["North", "South", "East", "West"],
        [0.25, 0.25, 0.25, 0.25]
    ))
    .add_column("promo_code", ("string", 6, "upper"))
    .add_column("name", ("names", True))
    .add_column("email", ("email",))
    .add_column("phone", ("phone", [813, 305, 212], "dashes"))
    .add_column("state", ("state", True))
    .add_column("country", ("country", "name", True))
    .add_column("signup_date", ("date", "2022-01-01", "2023-12-31"))
    .add_column("last_seen", ("datetime", "2023-01-01", "2024-01-01"))
    .add_column("dob", ("dob_from_age", "age"))
    .add_column("age_check", ("age_from_dob", "dob"))
    .add_column("income", ("lognormal", 10, 0.6, {"round": 2}))
    .add_column("noise_normal", ("normal", 0, 1))
    .add_column("bounded_score", ("normal_clip", 70, 10, 0, 100))
    .add_column("tickets_opened", ("poisson", 3))
    .add_column("probability", ("beta", 2, 5, 1))
    .add_column("wait_time", ("exponential", 5))
    .add_column("gamma_metric", ("gamma", 2, 3))
    .add_column("trend_value", ("trend", 100, 0.8, 5))
    .add_column("spend_estimate", ("correlated", "income", 0.02, 200))
    .add_column("vip_flag", ("conditional", "plan", "Premium", 1, 0))
    .add_column(
        "income_to_age_ratio",
        ("computed", lambda df: df["income"] / df["age"], ["income", "age"])
    )
    .add_column("income_roll_mean", ("rolling_mean", "income", 5))
    .add_column("income_roll_sum", ("rolling_sum", "income", 5))
    .add_column("income_roll_std", ("rolling_std", "income", 5))
    .add_column("income_roll_max", ("rolling_max", "income", 5))
    .add_column("income_roll_min", ("rolling_min", "income", 5))
    .add_column("income_z", ("zscore", "income"))
    .add_column(
        "income_bin",
        ("bin", "income", [0, 30000, 70000, 150000, 1_000_000],
        ["Low", "Mid", "High", "Very High"])
    )
    .add_column("income_pct_change", ("pct_change", "income"))
)

df = gen.build()
df.tail(10)
```

```python
# use case 2 (college)

from df_gen import df_gen
import numpy as np
import pandas as pd 

non_us_countries = [
    "Canada", "Mexico", "India", "China", "United Kingdom",
    "Germany", "France", "Brazil", "Nigeria", "South Korea"
]

gen = df_gen(10, random_state=10)
gen.add_column("id", ("prefix_id", "STU", 5, True))
gen.add_column("names", ("names",))
gen.add_column("student_email", ("email", ["school.edu"]))
gen.add_column("phone", ("phone",[813,833,767]))
gen.add_column("state", ("state", {"missing": 0.25}))
gen.add_column(
    "country",
    (
        "computed",
        lambda df: pd.Series(
            np.where(
                df["state"].notna(),
                "United States",
                np.random.choice(non_us_countries, size=len(df))
            ),
            dtype="object"
        ),
        ["state"]
    )
)
gen.add_column("age", ("int", 18, 22))
gen.add_column("dob", ("dob_from_age", "age"))
gen.add_column("major", ("category", ["art", "math", "music", "biology", "chemistry", "english", "political science", "psychology"]))
gen.add_column("GPA", ("normal_clip", 3,1.5, 0,4, {"round": 2})) #rounding used here
gen.add_column("major_GPA", ("decimals", 2, 4))
gen.add_column(
    "year",
    (
        "computed",
        lambda df: pd.Series(
            np.select(
                [
                    df["age"].between(18, 19),
                    df["age"].between(19, 20),
                    df["age"].between(20, 21),
                ],
                [
                    "Freshman",
                    "Sophomore",
                    "Junior",
                ],
                default="Senior"
            ),
            dtype="object"
        ),
        ["age"]
    )
)
df = gen.build()

df.head(10)
```
